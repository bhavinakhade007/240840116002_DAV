### AIM : Perform data inspection, cleaning, handling missing values, duplicate removal, and data transformation using Pandas.

**Step: 1 Import Pandas**

```python
# Python library used to work with data.
# It helps us read, organize, analyze, and process data easily.

import pandas as pd

# pd is the short name for the Pandas library.
```

**Step: 2 Create Dataframe**

```python
# Create sample data by storing example values in a dictionary.
data ={
    "Id":[101,102,103,104,104,105,106,107],  #student id
    "Name":["Rahul","Priya","Amit",None,"Riya","Ajay","Naresh","Jainil"],  #student name
    "Age":[20,21,None,20,20,22,23,28],  #student age
    "City":["Surat","Navsari","Surat","Bharuch","Bharuch","Navsari","Surat","Ahemedabad"],  #student city
    "Marks":[85,None,78,88,88,81,90,93]  #student marks
}
```

```python
df = pd.DataFrame(data)
df
```

**Step: 3 Inspect the Database**

```python
# Task 1: print only first 5 row of dataframe.
df.head()
```

```python
# Task 2: print only last 5 row of dataframe
df.tail()
```

```python
# Task 3: print detail information of dataframe
df.info()
```

```python
# Task 4: Generate statistical information.
df.describe()
```

```python
# Task 5: print dimension of dataframe.
df.shape
```

```python
# Task 6: print name of column.
df.columns
```

```python
# Task 7: print datatype of each column.
df.dtypes
```

**Step 4 : Check missing Values.**

```python
# Task 1: Check null values in each & every column.
df.isnull().sum()
```

**Step 5: Fill Missing Values.**

```python
# Task 1: Fill missing values (replace missing value in marks)

df["Marks"] = df["Marks"].fillna(df["Marks"].mean())

# print the Updated Dataframe.
df
```

```python
# Task 2: Fill missing Age.
# Replace missing values in the Age column.

df["Age"] = df["Age"].fillna(df["Age"].mean())

# print the Updated Dataframe.
df
```

```python
# Task 3: Fill missing Name.
# Replace missing name with the text "Unknown"

df["Name"] = df["Name"].fillna("Unknown")

# print the updated Dataframe.
df
```

**Step 6: Detect Duplicate Records.**

```python
# Task 1: check wheater each row is duplicated.

df.duplicated()
```

```python
# Task 2: Count total no of duplicate record.
df.duplicated().sum()
```

**Step 7: Remove Duplicate Records.**

```python
# Remove duplicate rows from the dataframe.
df = df.drop_duplicates()

# print the updated Dataframe.
df
```

**Step 8: Rename a column.**

```python
# Rename the column marks to "score".
# change directly to the original DataFrame using inplace=True.

df.rename(columns ={"Marks":"Score"},inplace=True)

# print the updated Dataframe.
df
```

**Step 9: Change Data type.**

```python
# Change the datatype.
#convert age column float -> integer.

df["Age"] = df["Age"].astype(int)
df.dtypes
```

**Step 10: Create a new column.**

```python
#Create a new column named "final score".

df["Final Score"] = df["Score"]+5

# print the updated Dataframe.
df
```

**Step 11: Convert text to uppercase.**

```python
# Convert all student name to uppercase.

df["Name"] = df["Name"].str.upper()

# print the updated Dataframe.
df
```

**Step 12: Convert text to lowercase.**

```python
#Convert city names to lowercase.
df["City"] = df["City"].str.lower()

# print the updated Dataframe.
df
```

**Step 13: Remove Extra spaces.**

```python
#Remove Extra (white)leading and trailing space from city name.

df["City"] = df["City"].str.strip()

# print the updated Dataframe.
df
```

**Step 14: Sort the Dataset.**

```python
# Sort data based on final score.
df_sorted = df.sort_values(by="Final Score",ascending=False)
df_sorted
```

**Step 15: Save the cleaned Dataset.**

```python
#Save clean dataframe as new CSV file.

df.to_csv("Cleaned_Students.CSV",index=False)
print("Dataset created successfully")
```
