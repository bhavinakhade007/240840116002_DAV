Apply ƒltering, sorting, aggregation, and GroupBy operations on real-world datasets.


```python
# Importing required libraries
# pandas -> for creating and manipulating dataframes
# numpy  -> for numerical operations

import pandas as pd
import numpy as np
# Creating a sample dataset in a dictionary
# Each key represents a column and its tuple value represents the column's data
data = {
    'emp_id': (101,102,103,104,105,106,107,108),
    'Name': ('rahul','priya','amit','riya','karan','neha','vivek','pooja'),
    'Department': ('HR','IT','Finance','HR','IT','Finance','HR','IT'),
    'Salary': (50000,60000,55000,70000,80000,52000,65000,75000),
    'City': ('Delhi','Mumbai','valsad','Mumbai','Delhi','navsari','surat','Mumbai'),
    'experience': (2,3,1,4,5,2,3,4)

}
```

step 1 : display the data set


```python

# Converting the dictionary into a pandas DataFrame
df = pd.DataFrame(data)
df
```

step 2: filter the data

```python
# Filtering employees who belong to the IT department
df[df["Department"]=="IT"]
```

employee salary grater than 50000

```python
#filtering employees who get salary above 50000
df[df["Salary"]>50000]
```

display emplooyee from surat

```python
#displaying employees from surat
df[df["City"]=="surat"]
```

APPLY MULTIPLE CONDITION

```python
# Applying multiple conditions together using the '&' (AND) operator
df[(df["Department"]=="IT") & (df["Salary"]>50000)]
```

sorting values by salary

```python
#sorting the values by salary in ascendig value
df.sort_values(by="Salary")
```

sorting the data


```python
#sorting the data by salary
df.sort_values(by="Salary",ascending=False)
```

sort by department and salary

```python
# First by Department (ascending), then by Salary (descending) within each department
df.sort_values(by=["Department","Salary"],ascending=[True,False])
```

step 4 : perfom aggregation
calculate total salary

```python
# Calculating the total (sum) of all salaries
df["Salary"].sum()
```

calculate average salary

```python
# Calculating the average (mean) salary of all employees
df["Salary"].mean()
```

find the highest salary

```python
# Finding the maximum salary among all employees
df["Salary"].max()
```

```python
#FIND THE LOWEST SALARY
df["Salary"].min()
```

```python
#count the employess
df["emp_id"].count()
```

step 5: perform groupby operation

```python
#avg salary department wise
df.groupby("Department")["Salary"].mean()
```

```python
#calculate minimum salary department wise
df.groupby("Department").min()
```

```python
#count employee department wise
df.groupby("Department")["emp_id"].count()
```

```python
# count total salary department wise
df.groupby("Department")["Salary"].sum()
```

```python
#count the maxx salary department wise
df.groupby("Department")["Salary"].max()
```

```python
#count minimum salary department wise
df.groupby("Department")["Salary"].min()
```

city wise avg salary


```python
# Grouping by City and calculating average salary per city
df.groupby("City")["Salary"].mean()
```

step 6: multiple aggeration function

```python
# Applying multiple aggregation functions at once (count, mean, max, min, sum)
df.groupby("Department")["Salary"].agg(["count","mean","max","min","sum"])
```

step 7: save the result

```python
#save data into a csv file .
df.to_csv("data.csv", index=False)
print("saved succesfuly")
```
