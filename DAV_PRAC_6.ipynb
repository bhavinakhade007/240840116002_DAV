step-1 : import library

```python
# Import the required libraries
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

Step-2 : create a dataset

```python
#create a dataset for the data
data = {
'Month': ['Jan', 'Feb', 'Mar', 'Apr', 'May', 'Jun', 'Jul', 'Aug', 'Sep', 'Oct', 'Nov', 'Dec'],
'product': ['Laptop', 'Mobile', 'Tablet', 'Laptop', 'Mobile', 'Tablet', 'Laptop', 'Mobile', 'Tablet','Laptop', 'Mobile', 'Tablet'],
'Region': ['north', 'south', 'east', 'west', 'north', 'south', 'east', 'west', 'north', 'south', 'east', 'west'],
'sales': [120000, 135000, 150000, 145000, 160000, 175000, 180000, 170000, 190000, 210000, 230000, 250000],
'units_sold' : [12, 30, 25, 14, 35, 30, 18, 38, 32, 21, 45, 40]
}
df = pd.DataFrame(data)
print(df)
```

step-3 : create a line chart for how did the company's sales changes monthly

```python
#craete simple momthly dataset
monthly_data = {
'Month': ['Jan', 'Feb', 'Mar', 'Apr', 'May', 'Jun', 'Jul', 'Aug', 'Sep', 'Oct', 'Nov', 'Dec'],
'sales': [120000, 135000, 150000, 145000, 160000, 175000, 180000, 170000, 190000, 210000, 230000, 250000],
'units_sold' : [12, 30, 25, 14, 35, 30, 18, 38, 32, 21, 45, 40]
}
df = pd.DataFrame(monthly_data)
print(df)

#create line chart of monthly data
plt.figure(figsize=(10, 5))
plt.plot(monthly_data["Month"], monthly_data["sales"],
marker="o", color="#1f77b4", label="Sales")
plt.title("Monthly Sales changes", fontsize=14)
plt.xlabel("Month")
plt.ylabel("Amount in $1000")
plt.legend()
plt.grid(True, alpha=0.3)
plt.tight_layout()


# Save chart
plt.savefig("line_chart.png", dpi=300)
plt.show()
```

step-4 : create bar chart for which product generates how much total sales

```python
#create prodect data set
product_data = {
'product': ['Laptop', 'Mobile', 'Tablet', 'Laptop', 'Mobile', 'Tablet', 'Laptop', 'Mobile', 'Tablet','Laptop', 'Mobile', 'Tablet'],
'sales': [120000, 135000, 150000, 145000, 160000, 175000, 180000, 170000, 190000, 210000, 230000, 250000],
'units_sold' : [12, 30, 25, 14, 35, 30, 18, 38, 32, 21, 45, 40]
}
df = pd.DataFrame(product_data)
print(df)

#craate bar graph for which product gebrates how much sales
plt.figure(figsize=(9, 5))
colors = plt.cm.viridis(np.linspace(0.2, 0.8, len(product_data)))
bars = plt.bar(product_data["product"],
product_data["units_sold"],
color=colors,
edgecolor="black")

plt.title("Units Sold by Product", fontsize=14)
plt.xlabel("Product")
plt.ylabel("Units Sold")
plt.tight_layout()
# Save chart
plt.savefig("bar_chart.png", dpi=300)
plt.show()
```

step -5 : create histogram how thw sales value is distributed

```python
# Create the histogram for the 'sales' column
plt.hist(df['sales'], bins=5, color='royalblue', edgecolor='black')

# Label your chart
plt.xlabel('Sales Value (INR)')
plt.ylabel('Frequency (Count of Months)')
plt.title('Distribution of Monthly Sales Values')

plt.show()
```

step-6 : create a scatter plot for relation between unit sold and total solds

```python
# Create the scatter plot
# We put the independent variable (units sold) on X, and dependent (sales) on Y
plt.scatter(df['units_sold'], df['sales'], color='darkorange', edgecolor='black', s=80)

# Label the axes and add a title
plt.xlabel('Number of Units Sold')
plt.ylabel('Total Sales Value')
plt.title('Relationship Between Units Sold and Total Sales')

# Display grid lines for better readability
plt.grid(True, linestyle='--', alpha=0.5)

plt.show()


```

step-7 : create pie chart for what percentage of total sales comes from each region

```python
data = {
'Month': ['Jan', 'Feb', 'Mar', 'Apr', 'May', 'Jun', 'Jul', 'Aug', 'Sep', 'Oct', 'Nov', 'Dec'],
'product': ['Laptop', 'Mobile', 'Tablet', 'Laptop', 'Mobile', 'Tablet', 'Laptop', 'Mobile', 'Tablet','Laptop', 'Mobile', 'Tablet'],
'Region': ['north', 'south', 'east', 'west', 'north', 'south', 'east', 'west', 'north', 'south', 'east', 'west'],
'sales': [120000, 135000, 150000, 145000, 160000, 175000, 180000, 170000, 190000, 210000, 230000, 250000],
'units_sold' : [12, 30, 25, 14, 35, 30, 18, 38, 32, 21, 45, 40]
}
df = pd.DataFrame(data)

# Step 1: Group by Region and sum the sales
region_sales = df.groupby('Region')['sales'].sum()

# Step 2: Define distinct colors for the slices
colors = ['#ff9999','#66b3ff','#99ff99','#ffcc99']

# Step 3: Create the pie chart
# autopct='%1.1f%%' automatically calculates and prints the percentage on each slice
plt.pie(region_sales,
        labels=region_sales.index,
        autopct='%1.1f%%',
        startangle=140,
        colors=colors,
        wedgeprops={'edgecolor': 'white', 'linewidth': 2})

# Give the chart a clean title
plt.title('Percentage of Total Sales by Region', fontsize=14, weight='bold')

plt.show()
```
