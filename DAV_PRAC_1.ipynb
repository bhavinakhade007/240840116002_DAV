### AIM : Study the Data Analytics lifecycle and perform data acquisition from CSV, Excel, and JSON ƒles using Pandas

**step 1: import the pandaas library**

```python
#Python library used to work with data.
# It helps to read, organize, analyze, and process data easily.

import pandas as pd

# pd is the short name for the pandas library.
```

**step 2: create sample data**

```python
# Create sample data for storing example values in a dictionary.
data = {
    "ID": [101,102,103,104,105,106,107],   # Student ID numbers
    "Name": ["Dev","rohan","meet","gourav","bhavin","rehan","dhaval"],    # Student names
    "Age" :[19,20,22,19,19,20,20],       # Student ages
    "City": ["Navsari","Surat","Navsari","Bharuch","Valsad","Bardoli","Surat"],  # Cities
    "Marks" :[78,88,90,75,69,94,89]                  # Student marks

}
```

**step 3 : create a data frame**

```python
# A DataFrame is a one type of data storing format in form of rows and columns.
# df is the variable that stores the DataFrame
df = pd.DataFrame(data)

# Display the DataFrame
df
```

**step 4: Save data as a csv file**

```python
# create a csv file
df.to_csv("students.csv", index=False)

# this massage give the conformation that the csv file created successfully.
print("csv file created successfully.")
```

**step 5 : save data as a excel file**

```python
# Save the DataFrame as an Excel file without adding the row index
# index=False means the row numbers will not be saved in the file.
df.to_excel("students.xlsx",index=False)

# this massage give the conformation that the excel file created successfully
print("Excel file created successfully.")
```

**Step 6 : Save data as JSON file**

```python
# Save the DataFrame as a JSON file with each row stored as a separate record
# orient="records" saves each row as a separate JSON object in a list
df.to_json("students.json", orient = "records" , index=4)

# this massage give the conformation that the JSON file created successfully
print("JSON file created successfully.")
```

**Step 7: Read the CSV file**

```python
# Read the data from the "students.csv" csv file and store it in a DataFrame
# pd.read_csv() -> Reads data from a CSV file
# df_csv ->  here the read data will be saved to the DataFrame

df_csv = pd.read_csv("students.csv")

# Display the DataFrame

df_csv
```

**Step 8:Read the Excel file**

```python
# Read the data from the "students.xlsx" Excel file and store it in a DataFrame
# pd.read_excel() -> Reads data from an Excel file
# df_excel ->  here the read data from the excel file will be saved to the DataFrame

df_excel = pd.read_excel("students.xlsx")

# Display the DataFrame

df_excel
```

**Step 9: Read the JSON file**

```python
# Read the data from the "students.json" JSON file and store it in a DataFrame
# pd.read_json() -> Reads data from a JSON file
# df_json ->  here the read data from the JSON file will be saved to the DataFrame

df_json = pd.read_json("students.json")

# Display the DataFrame

df_json
```
